## Nearest neighbour

```python
import os
import pandas as pd

csv_path = r"C:\Users\ewestlund\Documents\Python Projects\Roberto_MuSpAn\output_KNN10\nearestneighbour"

files = [f for f in os.listdir(csv_path) if f.endswith(".csv")]

columns_to_keep = ['object_id', 'x', 'y', 'object_type', 'Cell_id', 'Barcode', 'Cluster',
       'Compartment', 'Celltype', 'Class', 'Neighbourhood_ID_KNN_10',
       'Cancer_to_Fibroblast', 'Cancer_to_CD8 T cell',  'Cancer_to_T cell (Ctla4-high)',
       'Cancer_to_LYVE1+FOLR2+ resident macrophage', 'Cancer_to_Macrophage/monocyte', 'Cancer_to_Neutrophil', 'Cell centres']

columns_to_add = ['Cancer_to_Lymphoid', 'Cancer_to_Macrophage', 'Cancer_to_Myeloid']

for file in files:
    df = pd.read_csv(os.path.join(csv_path, file))

    cancer_df = df.loc[df["Celltype"] == "Cancer"]

    cancer_df = cancer_df[columns_to_keep]

    cancer_df.to_csv(os.path.join(csv_path, file.replace(".csv", "_cancerCellsOnly.csv")))
```
